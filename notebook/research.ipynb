```python
# Filtering only those users who read more than 200 times
# users with 1 or 2 or 10 ,,, it will consider as outliers
# Lets stores users who had at least rated  more than 200 books

x=ratings['user_id'].value_counts()>200
x
```

```python
ratings['user_id'].isin(y).value_counts()
```

```python
ratings=ratings[ratings['user_id'].isin(y)]
```

```python
books.head()
```

```python
# Now join the ratings and books based on ISBN

ratings_with_books=ratings.merge(books,on='ISBN')
```

```python
ratings_with_books.rename(columns={'book_rating':'rating'},inplace=True)
```

```python
ratings_with_books.head()
```

```python
ratings_with_books.shape
```

```python
number_rating=ratings_with_books.groupby('title')['rating'].count().reset_index()
```

```python
number_rating.rename(columns={'rating':'num_of_rating'},inplace=True)
```

```python
number_rating
```

```python
final_rating=ratings_with_books.merge(number_rating,on='title')
```
